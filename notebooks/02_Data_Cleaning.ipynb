```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/health_risk_dataset.csv")
```

```python
#create a copy

clean_df = df.copy()
```

```python
missing_values = clean_df.isnull().sum()
print("missing_values: ", missing_values)
```

```python
total_missing = clean_df.isnull().sum().sum()
print("Total missing values: ", total_missing)
```

```python
duplicate_count = clean_df.duplicated().sum()
print("Duplicate count: ", duplicate_count)
```

```python
clean_df.info()
```

```python
#check age
print(clean_df["Age"].describe())

print("Minimum age:", clean_df["Age"].min())
print("Maximum age:", clean_df["Age"].max())

#check BMI
print("Minimum BMI:", clean_df["bmi"].min())
print("Maximum BMI:", clean_df["bmi"].max())

#Check Blood Pressure
print("Minimum systolic BP:", clean_df["systolic_bp"].min())
print("Maximum systolic BP:", clean_df["systolic_bp"].max())

print("Minimum diastolic BP:", clean_df["diastolic_bp"].min())
print("Maximum diastolic BP:", clean_df["diastolic_bp"].max())
```

```python
#Check Categorical Values

print(clean_df["gender"].unique())

print(clean_df["smoking_status"].unique())

print(clean_df["employment_status"].unique())
```

```python
#Check Category Frequencies

print(clean_df["gender"].value_counts())

print(clean_df["smoking_status"].value_counts())

print(clean_df["employment_status"].value_counts())
```

```python
#Check Target Variables
target_columns = [
    "diabetes_risk_score",
    "hypertension_risk_score",
    "heart_disease_risk_score",
    "obesity_risk_score",
    "cholesterol_risk_score"
]

print(clean_df[target_columns].isnull().sum())
```

```python
#Check Target Ranges

for columns in target_columns:
    print(f"\n{columns}")
    print("Minimum:", clean_df[columns].min())
    print("Maximum:", clean_df[columns].max())
```

```python
print("Final number of rows:", clean_df.shape[0])
print("Final number of columns:", clean_df.shape[1])
```

```python
#save the cleaned dataset to a new CSV file
clean_df.to_csv(
    "../data/interim/clean_health_data.csv",
    index=False
)
```
