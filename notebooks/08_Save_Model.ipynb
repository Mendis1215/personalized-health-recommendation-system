# 08 - Save Final Models

## Objective

The purpose of this notebook is to save the best-performing machine
learning model for each health risk score based on the results from
the Model Evaluation notebook.

The selected models are:

- Diabetes Risk Score → Linear Regression
- Hypertension Risk Score → Ridge Regression
- Heart Disease Risk Score → Gradient Boosting
- Obesity Risk Score → Gradient Boosting
- Cholesterol Risk Score → Gradient Boosting

The final models will be saved for use in the prediction pipeline.

This notebook does not retrain the models.

It only selects and saves the already-trained models identified during
model evaluation.

```python
import pandas as pd
import numpy as np
import joblib
import os
```

```python
#Create final model directory

final_model_dir = "../models/final"

os.makedirs(
    final_model_dir,
    exist_ok=True
)

print("Final model directory:", final_model_dir)
```

```python
#Define the selected models

best_model_selection = {
    "diabetes_risk_score": "Linear Regression",
    "hypertension_risk_score": "Ridge Regression",
    "heart_disease_risk_score": "Gradient Boosting",
    "obesity_risk_score": "Gradient Boosting",
    "cholesterol_risk_score": "Gradient Boosting"
}

best_model_selection

#Display the model selection

print("Final Model Selection")

for target, model_name in best_model_selection.items():
    print(f"{target:30} → {model_name}")
```

```python
#Load candidate models

candidate_model_dir = "../models/candidates"

candidate_model_files = {
    "Linear Regression": "linear_regression.pkl",
    "Ridge Regression": "ridge_regression.pkl",
    "Random Forest": "random_forest.pkl",
    "Gradient Boosting": "gradient_boosting.pkl",
    "Extra Trees": "extra_trees.pkl"
}

#Load all candidate models

candidate_models = {}

for model_name, file_name in candidate_model_files.items():

    file_path = os.path.join(
        candidate_model_dir,
        file_name
    )

    candidate_models[model_name] = joblib.load(
        file_path
    )

    print(f"{model_name} loaded successfully.")

#Verify candidate models

print("\nAvailable Candidate Models")

for model_name in candidate_models:
    print( model_name)
```

```python
#Load evaluation results

evaluation_df = pd.read_csv(
    "../outputs/metrics/model_evaluation_results.csv"
)

evaluation_df.head()
```

```python
#Load best-model summary

best_models_df = pd.read_csv(
    "../outputs/metrics/best_models_by_target.csv",
    index_col=0
)

best_models_df
```

```python
#Verify selection against our expected results

for target, expected_model in best_model_selection.items():

    actual_model = best_models_df.loc[
        target,
        "Model"
    ]

    if actual_model != expected_model:

        raise ValueError(
            f"Model mismatch for {target}: "
            f"expected {expected_model}, "
            f"found {actual_model}"
        )

print("Model selection verified successfully.")
```

```python
#Understand the target indexes

target_columns = [
    "diabetes_risk_score",
    "hypertension_risk_score",
    "heart_disease_risk_score",
    "obesity_risk_score",
    "cholesterol_risk_score"
]

target_indices = {
    target: index
    for index, target in enumerate(target_columns)
}

target_indices
```

```python
#Save the final models

final_models = {}

for target, model_name in best_model_selection.items():

    model = candidate_models[model_name]

    final_models[target] = model

    print(
        f"{target} → {model_name}"
    )
```

```python
#Save each final model

for target, model in final_models.items():

    model_path = os.path.join(
        final_model_dir,
        f"{target.replace('_risk_score', '')}_model.pkl"
    )

    joblib.dump(
        model,
        model_path
    )

    print(
        f"Saved: {model_path}"
    )
```

```python
#Important: Save target/model mapping

final_model_info = {}

for target, model_name in best_model_selection.items():

    final_model_info[target] = {
        "model_name": model_name,
        "target_index": target_indices[target],
        "model_file": (
            f"{target.replace('_risk_score', '')}_model.pkl"
        )
    }

final_model_info
```

```python
#Save model information

joblib.dump(
    final_model_info,
    "../models/final_model_info.pkl"
)

print("Final model information saved successfully.")
```

```python
#preprocessing pipeline

preprocessing_path = "../models/preprocessing_pipeline.pkl"

if os.path.exists(preprocessing_path):

    print(
        "Preprocessing pipeline found successfully."
    )

else:

    print(
        "WARNING: preprocessing_pipeline.pkl was not found."
    )
```

```python
#Verify final models exist

print("Final Models")
print("=" * 50)

for target, info in final_model_info.items():

    path = os.path.join(
        final_model_dir,
        info["model_file"]
    )

    if os.path.exists(path):

        print(
            f"✓ {target}: {info['model_file']}"
        )

    else:

        print(
            f"✗ Missing: {target}"
        )
```

```python
#Load the saved final models

loaded_final_models = {}

for target, info in final_model_info.items():

    model_path = os.path.join(
        final_model_dir,
        info["model_file"]
    )

    loaded_final_models[target] = joblib.load(
        model_path
    )

    print(
        f"{target} loaded successfully."
    )

```

```python
#Verify model types

for target, model in loaded_final_models.items():

    print(
        f"{target}: {type(model).__name__}"
    )
```

```python
#Save a human-readable model summary

final_model_summary = best_models_df.copy()

final_model_summary.to_csv(
    "../outputs/metrics/final_model_selection.csv"
)

print(
    "Final model selection saved successfully."
)

#Final verification table

final_model_summary
```
